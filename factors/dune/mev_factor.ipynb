```python
import pandas as pd

cols = [
    'open_time', 'open', 'high', 'low', 'close', 'volume',
    'close_time', 'q_volume', 'trades', 'taker_base', 'taker_quote', 'ignore'
]

df = pd.read_csv(
    'data.csv', 
    names=cols, 
    index_col=False,
    dtype={'open': 'float32', 'high': 'float32', 'low': 'float32', 'close': 'float32', 'volume': 'float32'}
)

# 时间戳转换
df['open_time'] = pd.to_datetime(df['open_time'], unit='s') 
df.set_index('open_time', inplace=True)
df['net_buy_ratio'] = (2 * df['taker_base'] - df['volume']) / df['volume']
print(df.head(5))
df.to_parquet('data.parquet', index=True)
```

```python
import pandas as pd

# 假设 results.rows 是你拿到的那个 list
rows = query_result.result.rows  

# 1. 转化为 DataFrame
df_raw = pd.DataFrame(rows)

# 2. 透视表：将 user_type 转化为列，按时间聚合 volume
# 这样每一行就是一个唯一的 5 分钟桶
df_pivot = df_raw.pivot_table(
    index='time', 
    columns='user_type', 
    values='volume', 
    aggfunc='sum'
).fillna(0)

# 确保三列都存在（防止某些时段缺少某种类型）
for col in ['Other', 'Sandwich', 'Sandwiched']:
    if col not in df_pivot.columns:
        df_pivot[col] = 0

# 3. 计算核心因子
# total_volume: 该时段总成交
df_pivot['total_volume'] = df_pivot['Other'] + df_pivot['Sandwich'] + df_pivot['Sandwiched']

# factor_mev_intensity: 被夹金额占比 (LVR/摩擦代理指标)
df_pivot['mev_intensity'] = df_pivot['Sandwiched'] / df_pivot['total_volume']

# 整理时间索引
df_pivot.index = pd.to_datetime(df_pivot.index)
df_pivot = df_pivot.sort_index()

print(df_pivot)
```
